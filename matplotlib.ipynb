```python
import matplotlib.pyplot as plt
import numpy as np
```

#Line style Collors

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 25, 10])

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([2, 5, 6, 10, 3])

plt.plot(y, linestyle= 'dotted', color='red', marker='o')
plt.title("Basic Line Plot")
plt.show()
```

#Multiple Lines in One Graph Numpy and Motlotip

```python
x = np.array([1, 2, 3, 4, 7])
y1 = np.array([2, 5, 6, 10, 3])
y2 = np.array([3, 8, 1, 10, 5])

plt.plot(x, y1, label = 'Line1', color= 'red', marker = 's')
plt.plot(x, y2, label = 'Line2', color= 'green', linestyle='--', marker = 'o')
plt.title("Basic Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.legend()
plt.show()
```

**Bar Chart with Numpy and Maplotlib**

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color = "blue")
plt.title("Basic Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()

```

Scatter Plot with Numpy and Matplotip

```python
x  = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([5, 2, 4, 2, 1, 4, 5, 2])

plt.scatter(x, y, color = 'red')
plt.title("Basic Scatter Plot")
plt.xlabel("X ")
plt.ylabel("Y")
plt.show()
```

# Histogram with Numpy and Matplotlib

```python
data = np.array([3, 5, 6, 7, 8, 8, 9, 10])
plt.hist(data, color = 'green', bins=10, edgecolor='black')
plt.title("Basic Histogram")
plt.xlabel("Values")
plt.ylabel("Count")
plt.show()
```

# Sublots(Multiple Graphs Together)

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 5, 6, 10])
y2 = np.array([3, 8, 1, 10])
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color ="skyblue")
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.bar(x, y1, color ="red")
plt.title("Plot 2")

plt.tight_layout()
plt.show()


```
